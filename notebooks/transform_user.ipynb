### Height

```python
df["height_m"].min()
```

```python
df["height_m"].max()
```

```python
df["height_m"].isnull().sum()
```

```python
df["height_m"] = df.groupby("gender")["height_m"].transform(
    lambda x: x.fillna(x.median())
)
```

48 reported null values so we will use a metric to find out if they are male or female and then utilise the median value.

### Fat percentage

```python
df["fat_percentage"].min()
```

```python
df["fat_percentage"].max()
```

```python
df["fat_percentage"].isna().sum()
```

```python
df["fat_percentage"] = df.groupby(["gender", "experience_level"])["fat_percentage"].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
df["fat_percentage"].isna().sum()
```

This bases the fat percentage on the average of the gender and the level of experience. 

### Experience level

```python
df["experience_level"] = df["experience_level"].str.strip()
```

```python
df["experience_level"].unique()
```

```python
df["experience_level"] = (
    df["experience_level"]
    .str.upper()
    .replace({
        "MID": "INTERMEDIATE",
        "ADV": "ADVANCED",
        "BEGINER": "BEGINNER",
        "B": "BEGINNER",
        "INT": "INTERMEDIATE",
        "EXPERT": "ADVANCED"
    })
)



df["experience_level"].unique()
```

```python
df.isna().sum()
```

### Weight

```python
train = df[df["weight_kg"].notna()]
test = df[df["weight_kg"].isna()]

features = ["height_m", "age"]

model = LinearRegression()
model.fit(train[features], train["weight_kg"])

df.loc[df["weight_kg"].isna(), "weight_kg"] = model.predict(test[features])
```

```python
df["weight_kg"].isna().sum()
```
